## Part 2: Deriving the d-invariant of the associated 2-bridge link S(p,q)

### Question 1 (core): Write a function that calulates d(S(p, q), i).

The d invariant of S(p,q) is a collection of p rational numbers. We define ith index/term of the d variant (represented as a row vector) in accordance to the 3 rules provided in the prompt. For future reference, we shall define this as the function d_S(p,q,i), where p, q, i ∈ Z.

The following rules are:

• For any q and i, d(S(1,q),i) = 0;

• If a%p denotes the integer between 0 and p which is congruent to a mod p then d(S(p, q), i) = d(S(p, q%p), i%p);

• When p > q > 0 and 0 ≤ i < p+q,  d(S(p,q),i) = (2i + 1 − p − q)2 − pq − d(S(q,p),i).

We are given an example of a d-invariant d(S(9,7)), defined as:

d(S(9, 7)) = [0, −2/9, 4/9, 0, 4/9, −2/9, 0, −8/9, −8/9].

Therefore it is sensible to define i as the ith component of this vector, and we assume this throughout the question.

```python
from fractions import Fraction

# This function computes the d-invariant for a given S(p, q) at index i using fractions with memoization.
def d_S(p, q, i, memo={}):
   
    # Check if the result is already in the memoization dictionary
    if (p, q, i) in memo:
        return memo[(p, q, i)]
    
    # Define the base case: if p == 1, return 0 for any q and i
    if p == 1:
        return Fraction(0)
    
    # If p > q > 0 and 0 <= i < p + q, apply the recursive formula from the third rule given
    if p > q > 0 and 0 <= i < p + q:
        # Recursively calculate the value based on the given formula
        return (Fraction(2 * i + 1 - p - q) ** 2 - Fraction(p * q)) / (4 * p * q) - d_S(q, p, i, memo)
    else:
        # Modulo rule: recursively reduce p and i as the second rule states
        if q == 0 or p == 0:  # Consider the zero case seperately; If q or p are zero, recursion should end here.
            result = Fraction(0)
        else:
            result = d_S(p, q % p, i % p, memo)
    
    # Store the result in the memoization dictionary to avoid redundant calculations
    memo[(p, q, i)] = result
    return result
```

Now lets test if the modulo rule works: In the case of the given example d(S(9,7)) for i and j, where jmod9 = i, is d(S(9,7),i) and d(S(9,7),j) equal?

We can see in the given example d(S(9,7)) that the module rule indeed works.

```python
##TESTING CELL

def check_modulo_rule(p, q, k):
    # Check the modulo rule for d(S(p, q)). We need to show for i, j where j % p == i, assert that d(S(p, q), i) == d(S(p, q), j).
    
    # First compute d-invariants for all i < p
    for i in range(p):
        for j in range(p, k):  # Check for all pairs where p <= j < k
            if j % p == i:  # Check if j mod p == i
                # Compute the d-invariants for S(p, q), i and j
                di = d_S(p, q, i)
                dj = d_S(p, q, j)
                assert di == dj, f"Modulo rule fails: d(S({p},{q}), {i}) = {di} != d(S({p},{q}), {j}) = {dj}"
                print(f"Modulo rule works: d(S({p},{q}), {i}) == d(S({p},{q}), {j}) = {di}")


# Using the example given:
p = 9
q = 7

# For k can input any integer here where j is greater than p and less than k. We can see for this example that the modulo rule works for all j greater than 8.

print("Checking modulo rule for d(S(9,7),i) where 0 ≤ i ≤ 20.")
print("------------------------------------------------------------------------------------------")
check_modulo_rule(p, q, 20)
print("------------------------------------------------------------------------------------------")
print("Checking modulo rule for d(S(9,7),i) where 0 ≤ i ≤ 30.")
print("------------------------------------------------------------------------------------------")
check_modulo_rule(p, q, 30)
```

### Question 2 (core): Write a function that computes the list of d invariants for S(p,q).

Here all we have to do is take all the ith components represented as d(S(p,q),i), where 0 ≤ i ≤ p-1, and present it as a row vector. For future reference, we shall define this as the function D_S(p,q), where p, q ∈ Z, p > 0. 

```python
def D_S(p, q):
    # This computes the list of d-invariants for S(p, q) for all i in the range [0, p-1].
    return [(d_S(p, q, i)) for i in range(p)]
```

### Question 3 (extension): For examples where you know S(p,q1) = S(p,q2) determine how this is reflected in the d-invariant.

In cases where you know S(p,q1) = S(p,q2) we can determine how this is reflected in the d-invariant, by taking an example of
where S(p,q1) = S(p,q2) and finding their d-invariants, and comparing it to an example of when S(p,q1) =\ S(p,q2) and 
checking their d-invariants.

We can take our pairs to be S(7,1) and S(7,2), and S(7,1) and S(7,8). Then, we need to check which of the pairs statisfies the
equation S(p,q1) = S(p,q2), and we do this by checking each pair to see if they statisfy one of these two conditions:

• q1 = q2 (mod p)

• q1q2 = 1 (mod p)

(page 492, P.Lisca, Lens spaces, rational balls and the ribbon conjecture)

We will now check these pairs below:


```python

pair1_q1 = 1
pair1_q2 = 2
pair2_q1 = 1
pair2_q2 = 8
p = 7

#Check if the first condition is valid for pair 1
modpair1_q2 = pair1_q2 % p
if modpair1_q2 == pair1_q1:
    print("The first condtion is valid for pair 1, hence S(7,1) = S(7,2)")
else:
    print("The first condition isn't valid for pair 1")
    #Check if the second condition is valid for pair 2
    q1q2 = pair1_q1 * pair1_q2
    if q1q2 == pair1_q1 % p:
        print("The second condition is valid for pair 1, hence S(7,1) = S(7,2)")
    else:
        print("Neither condtion is valid for pair 1, hence S(7,1) =\ S(7,2)")
#Check if the first condition is valid for pair 2
modpair2_q2 = pair2_q2 % p
if modpair2_q2 == pair2_q1:
    print("The first condtion is valid for pair 2, hence S(7,1) = S(7,8)")
else:
    print("The first condition isn't valid for pair 2")
    #Check if the second condition is valid for pair 2
    q1q2 = pair2_q1 * pair2_q2
    if q1q2 == pair2_q1 % p:
        print("The second condition is valid for pair 2, hence S(7,1) = S(7,8)")
    else:
        print("Neither condtion is valid for pair 2, hence S(7,1) =\ S(7,8)")
```

From the code above we can see that pair 2 statisfies one of the conditions, and pair 1 statifies neither condition, therefore
pair 2 is valid for the equation S(p,q1) = S(p,q2), while pair 1 doesn't statisfy this equation. Hence, we can now find the 
d-invariants of all these pairs and compare our results to answer the intial question.

For this piece of code we can use the function previously created to compute the list of d-invariants for S(p,q):

```python
print("D-invariants for S(7,1): ")
print(D_S(7,1))
print("D-invariants for S(7,2): ")
print(D_S(7,2))
print("D-invariants for S(7,8): ")
print(D_S(7,8))
```

Hence, by looking at the results we can see that for pair 2 they have the same d-invariants, while for pair 1 the each part 
of the pairs have d-invariants. Therefore, as pair 2 statisfies the equation S(p,q1) = S(p,q2), while pair 1 doesn't, we can 
determine that S(p,q1) = S(p,q2) is reflected in d-invariant by having the same comoputed list of d-invariants for both parts 
of the pair. Hence the equation holds in the d-invariant as well.

### Question 4 (core): Does it completely distinguish between 2-bridge links?

Now we need to find whether the function D_S(p,q) completely distinguishes between 2-bridge links. 

In other words, we need to assess whether it is true that whenever S(p, q1) and S(p, q2) are inequivalent links they have differing d invariants. 
We will be checking this for a collection of varying sizes, such as when p < 100, and larger collections such as p < 1000.

We are using a counter method to easily determine whether a d-invariant is found more than once in any given collection; finding a duplicate is enough to show this.

Say the size of the collection is the integer max_p.

For all 2 ≤ p ≤ max_p, p > q > 0, we are storing all D_S(p,q) in a tuple and defining the dictionary seen_invariants to store previously encountered lists of D_S(p,q)).
If a tuple is seen before, the counter will be adjusted by 1 and will be flagged as a duplicate.

If the counter returns as zero, we can conclude that d(S(p,q)) is unique for S(p,q).

```python
def is_distinguished(max_p):
    # Compare d-invariants for all 2-bridge links with p < max_p to check if they distinguish inequivalent links.
    seen_invariants = {}
    counter = 0
    
    # Iterate over possible values of p and q for 2-bridge links. Note that p > q > 0 has to hold.
    for p in range(2, max_p):
        for q in range(1, p):
            d_inv = tuple(D_S(p, q))  # Compute the d-invariant list for S(p, q)
            if d_inv in seen_invariants:
                print(f"Found duplicate d-invariant for S({p}, {q}) and S{seen_invariants[d_inv]}")
                counter += 1
            else:
                seen_invariants[d_inv] = (p, q)
    
    return counter
```

```python
##TESTING CELL

print("For a collection of 100:", is_distinguished(100))
print("---------------------------------------------------------")
print("For a collection of 200:", is_distinguished(200))
```

This is to avoid performance degradation for larger values of p, as the recursive function will repeatedly compute the same values.
